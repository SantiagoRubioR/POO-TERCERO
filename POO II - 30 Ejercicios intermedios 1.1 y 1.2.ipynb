### Ejercicio 1 · Factura con descuento e IVA ★★

Pida el nombre de un producto, la cantidad y el precio unitario. Calcule:
- Subtotal = cantidad × precio
- Descuento del 10 % sobre el subtotal
- IVA del 15 % sobre (subtotal − descuento)
- Total

Muestre la factura con los valores **alineados a la derecha** en 10 espacios y 2 decimales:
```
Producto: Teclado
Subtotal:      90.00
Descuento:      9.00
IVA:           12.15
TOTAL:         93.15
```

````python
# Escriba su solución aquí
producto = input("Ingrese el producto: ")
cantidad = int(input("Ingrese la cantidad: "))
precio = float(input("Ingrese el precio unitario: "))

subtotal = cantidad * precio
descuento = subtotal * 0.10
iva = (subtotal - descuento) * 0.15
total = subtotal - descuento + iva

print(f"Producto: {producto:>10}")
print(f"Subtotal: {subtotal:>10.2f}")
print(f"Descuento: {descuento:>10.2f}")
print(f"IVA:       {iva:>10.2f}")
print(f"TOTAL:     {total:>10.2f}")
````

````python
# Escriba su solución aquí
capital = float(input("Ingrese la capital inicial: "))
tasa = float(input("Ingrese la tasa anual (%): "))
anios = int(input("Ingrese los años de interes: "))

monto_compuesto = capital * (1 + (tasa / 100)) ** anios
ganancia_compuesta = monto_compuesto - capital
interes_simple = capital * (tasa / 100) * anios
diferencia = ganancia_compuesta - interes_simple

print(f"Monto final compuesto: {monto_compuesto:,.2f}$")
print(f"Ganancia compuesta: {ganancia_compuesta:,.2f}$")
print(f"Diferencia con interes simple: {diferencia:,.2f}$")
````

### Ejercicio 22 · Inventario con diccionario ★★

Con `stock = {"laptop": 12, "mouse": 45, "teclado": 30, "monitor": 8}`:
1. Se vendieron 5 mouses: actualice el stock.
2. Llegó un producto nuevo: agregue `"webcam": 15` con `update`.
3. Se dejó de vender el monitor: elimínelo con `pop` y muestre cuántas unidades tenía.
4. Consulte `"impresora"` con `get` para que muestre `0` si no existe.
5. Muestre el total de unidades, los productos en orden alfabético y el producto con más unidades (`max(stock, key=stock.get)`).

````python
stock = {"laptop": 12, "mouse": 45, "teclado": 30, "monitor": 8}
# Escriba su solución aquí

stock["mouse"] -= 5
stock.update({"webcam": 15})
eliminado = stock.pop("monitor")
impresora = stock.get("impresora",0)

total_unidades = sum(stock.values())
alfabetico = sorted(stock.keys())
mas_unidades = max(stock, key=stock.get)

print(f"Monitor eliminado tenía: {eliminado} unidades")
print(f"Impresora: {impresora}")
print(f"Total unidades: {total_unidades}")
print(f"Orden alfabético: {alfabetico}")
print(f"Más unidades: {mas_unidades}")
````

```python
excel_t  = {"Luis", "Carla", "Pedro", "Diego"}
ia_t     = {"Ana", "Pedro", "Diego"}
```
Muestre (ordenado con `sorted`):
1. Quiénes están en Python **y** en Excel.
2. Quiénes están en los tres talleres.
3. Quiénes están en Python pero **no** en IA.
4. Quiénes están en Python **o** en Excel, pero no en ambos (`^`).
5. Si todos los de IA están también en Python (`issubset`).
6. El total de estudiantes distintos.

````python
python_t = {"Ana", "Luis", "María", "Pedro", "Sofía"}
excel_t = {"Luis", "Carla", "Pedro", "Diego"}
ia_t = {"Ana", "Pedro", "Diego"}
# Escriba su solución aquí

python_excel = sorted(python_t & excel_t)
todos = sorted(python_t & excel_t & ia_t)
python_no_ia = sorted(python_t - ia_t)
solo_uno = sorted(python_t ^ excel_t)
ia_en_py = ia_t.issubset(python_t)
total_estudiantes = len(python_t | excel_t | ia_t)

print(f"Python y Excel: {python_excel}")
print(f"En los tres: {todos}")
print(f"Python sin IA: {python_no_ia}")
print(f"Python o Excel (no ambos): {solo_uno}")
print(f"Todos de IA en Python: {ia_en_py}")
print(f"Total estudiantes: {total_estudiantes}")
````

### Ejercicio 30 · Ficha de un socio a partir de texto ★★

Un archivo antiguo guarda cada socio en una sola línea:
`linea = "ana pérez|0104567896|ahorros=1520.75|plazo fijo=5000"`
1. Separe la línea por `"|"`.
2. Cree un diccionario con `nombre` (formato título), `cedula` y `cuentas`, que a su vez es un diccionario `{tipo: saldo}` (separe cada cuenta por `"="` y convierta el saldo a `float`).
3. Muestre la ficha con este formato:
```
Socio: Ana Pérez (0104567896)
Tipos de cuenta: ahorros, plazo fijo
Saldo total: $6,520.75
```

````python
linea = "ana pérez|0104567896|ahorros=1520.75|plazo fijo=5000"
# Escriba su solución aquí

partes = linea.split("|")
tipo1, saldo1 = partes[2].split("=")
tipo2, saldo2 = partes[3].split("=")
diccionario = {
    "nombre": partes[0].title(),
    "cedula": partes[1], 
    "cuentas": {
        tipo1: float(saldo1),
        tipo2: float(saldo2)
    }
}

saldo_total = sum(diccionario["cuentas"].values())
tipos_cuenta = ", ".join(diccionario["cuentas"].keys())

print(f"Socio: {diccionario['nombre']} ({diccionario['cedula']})")
print(f"Tipos de cuenta: {tipos_cuenta}")
print(f"Saldo total: ${saldo_total:,.2f}")
````
